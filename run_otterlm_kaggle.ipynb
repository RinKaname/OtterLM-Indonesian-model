### 4. Launch Distributed Training (DDP)
This is the magic command. Instead of using Python directly or `DataParallel`, we use `torchrun`. 

Setting `--nproc_per_node=2` tells PyTorch to launch **two separate processes**, one for each T4 GPU. The `train.py` script is already written to handle DistributedDataParallel (DDP), so it will automatically sync gradients across the two GPUs.

```python
# Ensure you have selected 'GPU T4 x2' in your Kaggle Notebook settings!
# We use 'cd OtterLM &&' so the bash shell runs this in the correct cloned folder.
!cd OtterLM && torchrun --standalone --nproc_per_node=2 train.py
```
